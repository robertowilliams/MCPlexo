```python
import string
import secrets
import hashlib
import base64

# Allowed characters: A-Z, a-z, 0-9
ALPHABET = string.ascii_letters + string.digits


def generate_short_id(length=6, deterministic=False, seed: str | None = None):
    """
    Generate a short unique ID.
    
    Args:
        length (int): Number of characters in the ID (e.g. 5 or 6)
        deterministic (bool): If True, derive ID deterministically from `seed`
        seed (str | None): Text to derive deterministic ID from

    Returns:
        str: A short alphanumeric ID
    """
    if deterministic:
        if not seed:
            raise ValueError("A seed string is required for deterministic mode.")
        # Hash the seed with SHA1 → base64 encode → sanitize and trim
        h = hashlib.sha1(seed.encode()).digest()
        b64 = base64.b64encode(h).decode()
        # Replace URL-unsafe chars and restrict to ALPHABET range
        b64 = b64.replace('+', 'A').replace('/', 'B').replace('=', '')
        return b64[:length]
    else:
        # Random cryptographically secure ID
        return ''.join(secrets.choice(ALPHABET) for _ in range(length))


# 🧩 Example usage
if __name__ == "__main__":
    print("Random 5-char ID:", generate_short_id(5))
    print("Random 6-char ID:", generate_short_id(6))
    
    print("Deterministic 5-char (seed='hello'):", generate_short_id(5, deterministic=True, seed="hellohellohellohellohellohello"))
    print("Deterministic 6-char (seed='world'):", generate_short_id(6, deterministic=True, seed="world"))
```
